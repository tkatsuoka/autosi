```python
# 10000iterで一様分布からサンプリングしたhistogram
import numpy as np
from matplotlib import pyplot as plt

rng = np.random.default_rng(3)
p_values = []
n_iterations = 10000
for i in range(n_iterations):
    p_value = rng.uniform(0, 1)
    p_values.append(p_value)
    if (i + 1) % 1000 == 0:
        print(f"Iteration {i + 1}/{n_iterations} completed.")

plt.hist(p_values, bins=20, range=(0, 1))
plt.xlabel("p-value")
plt.ylabel("Frequency")
plt.show()
```
